## 5-Minute Demo Outline
1. **Question:** How can we predict organic Click-Through Rate (CTR) and post-click engagement using pre-publication features to solve the FlyRank content problem?
2. **Method:** Evaluated Ridge Regression against baseline and Random Forest models with leakage-free feature scaling.
3. **One Chart:** Horizontal bar chart of Feature Importances showing `reading_time` and `word_count` as top positive drivers, and `num_images` as a negative driver.
4. **One Honest Result:** Ridge regression achieved an $R^2$ of 0.63 (outperforming the 0.41 baseline), though accuracy drops on sparse, low-volume text categories.
