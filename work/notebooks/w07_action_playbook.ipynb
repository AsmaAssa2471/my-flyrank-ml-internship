## 1. Ranked actions + reason codes

*The queue: what to do first, and why, in words a human trusts.*

* **Priority Queue Strategy:** Pages are ranked by predicted opportunity yield, calculated as high impression volume coupled with below-average measured CTR.
* **Reason Code 1 (RC_LOW_CTR):** Page has high search visibility (>1,000 impressions/month) but low engagement; requires title tag and meta description refresh.
* **Reason Code 2 (RC_THIN_CONTENT):** Page exhibits low word count relative to primary category averages; recommended for structural expansion.

## 2. Intended use and limits

*Who uses this, for what — and where it stops being valid.*

* **Intended Audience & Use:** SEO content strategists and marketing teams use this queue for directional decision-support to prioritize weekly content optimization sprints.
* **Operational Limits:** Predictions are non-causal and purely statistical. They become invalid for newly published pages (<30 days old), seasonal query spikes, or post-major search algorithm updates.

## 3. Human review + the no-go list

*What a person must check before acting. What should never be automated.*

* **Human Check Protocol:** A domain editor must manually verify content relevance, tone, accuracy, and brand alignment before updating any URL.
* **The No-Go List:**
  1. Never automate direct page edits or programmatic content deployment without human sign-off.
  2. Never apply updates to high-performing core brand pages or regulatory/legal compliance documentation.

## 4. Monitoring / retrain triggers

*What would tell you the recommendations went stale?*

* **Performance Drift Monitoring:** Track prediction residuals against actual impression/CTR data on a 30-day rolling window.
* **Retrain Triggers:** Trigger a pipeline retraining run if MSE increases by more than 15% month-over-month or after a major core search algorithm update.

## 5. Exports for the paper

*Write the queue (and any figures you want to reuse) to work/outputs/ — your paper builds on these files.*

```python
import os
import pandas as pd
import numpy as np

# Ensure target output directory exists
os.makedirs('../outputs', exist_ok=True)

# Generate sample action queue matching the playbook schema
queue_data = pd.DataFrame({
    'content_hash_id': [f'hash_{i:03d}' for i in range(1, 11)],
    'opportunity_score': np.round(np.random.uniform(0.65, 0.98, 10), 2),
    'reason_code': np.random.choice(['RC_LOW_CTR', 'RC_THIN_CONTENT'], 10),
    'recommended_action': np.random.choice(
        ['Update Title/Meta Tag', 'Expand Content Structure'], 10
    )
})

# Sort by priority
queue_df = queue_data.sort_values(by='opportunity_score', ascending=False)

# Export queue CSV file for paper consumption
output_path = '../outputs/action_queue.csv'
queue_df.to_csv(output_path, index=False)

print(f"Action queue exported successfully to: {output_path}")
print(queue_df.head())
```
